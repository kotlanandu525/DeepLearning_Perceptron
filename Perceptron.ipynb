## **Perception Binary Classification**
ex: study hours-> Pass(1),Fail(0

**Step1:Prepare Trainin Data**

```python
#Training data
import numpy as np
import matplotlib.pyplot as plt
x=np.array([1,2,3,4])
y=np.array([0,0,1,1])
x,y
```

Step2:initialize the parameters

```python
w=0.0
b=0.0
learning_rate=1
epochs=5

print("initial weight",w)
print("initial bias",b)
```

**Step3:define activation function**

```python
def step_function(z):
  return 1 if z>-0 else 0
```

**Training loop**

```python
for epoch in range(epochs):
  print("\nEpoch",epoch+1)
  for i in range(len(x)):
    #1.compute weighted sum
    z=w*x[i]+b
    #2.Apply Activation fxn
    y_pred=step_function(z)
    #3.calculate error
    error=y[i]-y_pred
    #4.update weight
    w=w+learning_rate*error*x[i]
    #5.update bias
    b=b+learning_rate*error
    print(f"Input:{x[i]} ,Predicted:{y_pred}  ,Actual value:{y[i]} ,w:{w}  ,b:{b}")

```

**Final model Parameters**

```python
print("Final weight",w)
print("Final bias",b)
```

**Testing Model**

```python
test_val=3.0
z=w*test_val+b
prediction=step_function(z)
print("for study hours",test_val,", Prediction",prediction)
```

# **Email spam or not**

```python
# Simple Perceptron for Email Spam Detection (OR problem)

import numpy as np
import matplotlib.pyplot as plt

# Training data
# x1 = Suspicious Keywords, x2 = Unknown Sender
X = np.array([
    [1, 1],
    [1, 0],
    [0, 1],
    [0, 0]
])

# Labels (Spam = 1, Not Spam = 0) -> OR function
y = np.array([1, 1, 1, 0])

# Initialize weights and bias
w = np.zeros(2)
b = 0.0
lr = 0.1  # learning rate

# Step function
def step(z):
    return 1 if z >= 0 else 0

# Training loop
epochs = 10
for epoch in range(epochs):
    total_error = 0
    for i in range(len(X)):
        z = np.dot(w, X[i]) + b
        y_pred = step(z)
        error = y[i] - y_pred

        # Update rule
        w = w + lr * error * X[i]
        b = b + lr * error

        total_error += abs(error)

    print(f"Epoch {epoch+1}: weights = {w}, bias = {b}, total_error = {total_error}")

print("\nFinal weights:", w)
print("Final bias:", b)

# Plot the data points
plt.figure()
for i in range(len(X)):
    if y[i] == 1:
        plt.scatter(X[i,0], X[i,1], marker='o', label='Spam' if i==0 else "")
    else:
        plt.scatter(X[i,0], X[i,1], marker='x', label='Not Spam')

# Decision boundary: w1*x1 + w2*x2 + b = 0  -> x2 = -(w1*x1 + b)/w2
x_vals = np.linspace(-0.5, 1.5, 100)
if w[1] != 0:
    y_vals = -(w[0]*x_vals + b) / w[1]
    plt.plot(x_vals, y_vals, label="Decision Boundary")

plt.xlabel("x1 (Suspicious Keywords)")
plt.ylabel("x2 (Unknown Sender)")
plt.title("Perceptron Decision Boundary for Spam Detection")
plt.legend()
plt.grid(True)
plt.show()
```
